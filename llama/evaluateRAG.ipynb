```python
dataset = [

    EvaluationSample(
        query="Como o COSO define Enterprise Risk Management?",
        relevant_chunks=[
            """Enterprise risk management is de ned here as:

 The culture, capabilities, and practices, integrated with strategy-
 setting and performance, that organizations rely on to manage risk
 in creating, preserving, and realizing value."""
        ]
    ),

    EvaluationSample(
        query="Quantos componentes estruturam o framework?",
        relevant_chunks=[
            "The Framework consists of the ve interrelated components of enterprise risk management."
        ]
    ),

    EvaluationSample(
        query="Quais são os cinco componentes do framework de ERM?",
        relevant_chunks=[
            """The ve components11 are:

• Governance and Culture
• Strategy and Objective-Setting
• Performance
• Review and Revision
• Information, Communication, and Reporting"""
        ]
    ),

    EvaluationSample(
        query="O ERM deve ser integrado à estratégia?",
        relevant_chunks=[
            """• Strategy and Objective-Setting: Enterprise risk management is
  integrated into the entity’s strategic plan through the process of
  setting strategy and business objectives."""
        ]
    ),

    EvaluationSample(
        query="O ERM é um departamento separado?",
        relevant_chunks=[
            """Enterprise risk management is not static, nor is it an adjunct to a
business. Rather, it is continually applied to the entire scope of
activities as well as special projects and new initiatives."""
        ]
    ),

    EvaluationSample(
        query="Quem supervisiona o Enterprise Risk Management?",
        relevant_chunks=[
           """• Review and Revision: By reviewing enterprise risk management
  capabilities and practices, and the entity’s performance relative to
  its targets, an organization can consider how well the enterprise
  risk management capabilities and practices have increased value
  over time and will continue to drive value in light of substantial
  changes."""
        ]
    ),

    EvaluationSample(
        query="Qual é o objetivo do ERM em relação ao valor?",
        relevant_chunks=[
            "Enterprise risk management is de ned here as:\n\n"
            " The culture, capabilities, and practices, integrated with strategy-\n"
            " setting and performance, that organizations rely on to manage risk\n"
            " in creating, preserving, and realizing value.\n\n"
            "Enterprise risk management practices help the\n"
            "organization identify, prioritize, and focus on those risks that may\n"
            "prevent value from being created, preserved, and realized, or that\n"
            "may erode existing value. But, just as important, it also helps the\n"
            "organization pursue potential opportunities."
        ]
    ),


    EvaluationSample(
        query="O ERM reduz surpresas organizacionais?",
        relevant_chunks=[
            "Increasing positive outcomes and advantage while reducing\n"
            "  negative surprises: Enterprise risk management allows entities to\n"
            "  improve their ability to identify risks and establish appropriate\n"
            "  responses, reducing surprises and related costs or losses, while\n"
            "  pro ting from advantageous developments.\n\n"
            "Enterprise risk management allows organizations to anticipate the risks that would\n"
            "affect performance and enable them to put in place the actions needed to minimize\n"
            "disruption and maximize opportunity."
        ]
    ),

    EvaluationSample(
        query="A cultura faz parte do ERM?",
        relevant_chunks=[
            "Culture is developed and shaped by the people at all levels of an\n"
            "entity by what they say and do. It is people who establish the\n"
            "entity’s mission, strategy, and business objectives, and put\n"
            "enterprise risk management practices in place.\n\n"
            "Governance and culture together form\n"
            "  a basis for all other components of enterprise risk management."
        ]
    ),

    EvaluationSample(
        query="Como o ERM contribui para o alinhamento entre performance e risco?",
        relevant_chunks=[
            "It is integrated into\n"
            "the development of strategy, formulation of business objectives, and\n"
            "the implementation of those objectives through day-to-day\n"
            "decision-making.\n\n"
            "is integrated across strategy development, business objective\n"
            "formulation, and implementation and performance, it can enhance\n"
            "value."
        ]
    ),

    EvaluationSample(
        query="Qual é a relação entre ERM e resiliência organizacional?",
        relevant_chunks=[
            "Enhancing enterprise resilience: An entity’s medium- and long-\n"
            "  term viability depends on its ability to anticipate and respond to\n"
            "  change, not only to survive but also to evolve and thrive.\n\n"
            "The capacity to adapt to change makes an organization more\n"
            "resilient and better able to evolve in the face of marketplace and\n"
            "resource constraints."
        ]
    ),

    EvaluationSample(
        query="O ERM ajuda a identificar oportunidades além de riscos?",
        relevant_chunks=[
            "Organizations commonly focus on those risks that may result in a\n"
            "negative outcome, such as damage from a re, losing a key\n"
            "customer, or a new competitor emerging. However, events can also\n"
            "have positive outcomes,8 such as better-than-forecast weather,\n"
            "stronger staff retention trends, or improved tax rates, which should\n"
            "also be considered.\n\n"
            "By considering all possibilities—both positive and negative aspects of risk—\n"
            "  management can identify new opportunities and unique\n"
            "  challenges associated with current opportunities."
        ]
    ),

    EvaluationSample(
        query="O que é risk appetite no contexto do framework?",
        relevant_chunks=[
            "Management is able to make decisions that are appropriate given\n"
            "  its appetite, risk pro le of the entity, and the changes to the pro le\n"
            "  that occur over time.\n\n"
            "The practices are intended to help people within the entity better understand\n"
            "what the acceptable amount of risk is, how risk impacts performance."
        ]
    ),

    EvaluationSample(
        query="Como os riscos são tratados no componente Performance?",
        relevant_chunks=[
            "Management actively addresses risk in pursuit of its performance\n"
            "  targets.\n\n"
            "Enterprise risk management allows organizations to anticipate the risks that would affect\n"
            "performance and enable them to take action to minimize disruption."
        ]
    ),

    EvaluationSample(
        query="Qual o papel do componente Review and Revision?",
        relevant_chunks=[
            "Review and Revision represent\n"
            " the common processes that ow through the entity.\n\n"
            "By reviewing enterprise risk management capabilities and practices, and the entity’s\n"
            "performance relative to its targets, an organization can consider how well enterprise\n"
            "risk management has increased value over time."
        ]
    ),

    EvaluationSample(
        query="Qual o papel da comunicação no ERM?",
        relevant_chunks=[
            "Information, Communication,\n"
            "and Reporting represent supporting aspects of enterprise risk\n"
            "management.\n\n"
            "Communication is the continual, iterative process of obtaining\n"
            "information and sharing it throughout the entity."
        ]
    ),

    EvaluationSample(
        query="Como o ERM auxilia na alocação de recursos?",
        relevant_chunks=[
            "Obtaining robust information on risk allows\n"
            "  management, in the face of nite resources, to assess overall\n"
            "  resource needs, prioritize resource deployment and enhance\n"
            "  resource allocation.\n\n"
            "Every risk could be considered a\n"
            "  request for resources."
        ]
    ),

    EvaluationSample(
        query="Como o ERM ajuda a gerenciar riscos em toda a entidade?",
        relevant_chunks=[
            "Identifying and managing risk entity-wide: Every entity faces\n"
            "  myriad risks that can affect many parts of the organization.\n"
            "  Sometimes a risk can originate in one part of the entity but impact\n"
            "  a different part.\n\n"
            "Management identi es and manages these entity-wide risks to\n"
            "sustain and improve performance."
        ]
    ),

    EvaluationSample(
        query="Qual é a definição de risco segundo o framework?",
        relevant_chunks=[
            "risk is de ned as:\n"
            "   The possibility that events will occur and affect the achievement of\n"
            "   strategy and business objectives.\n\n"
            "  • Uncertainty: The state of not knowing how or if potential\n"
            "    events may manifest.\n"
            "  • Severity: A measurement of considerations such as the\n"
            "    likelihood and impact of events or the time it takes to\n"
            "    recover from events."
        ]
    ),

 ]





# =====================================================
# PIPELINE REAL (Chroma + RankContext)
# =====================================================

def build_pipeline():

    from langchain_chroma import Chroma
    from langchain_huggingface import HuggingFaceEmbeddings
    from RAG.rank_context import RankContext
    from RAG.BuildVectorStore import PDFVectorStore
    rag_support = PDFVectorStore(persist_dir='RAG/db')

    chroma = rag_support._load_db()

    reranker = RankContext(
        model="BAAI/bge-reranker-large",
        top_n=6
    )

    def pipeline(query: str, k: int):

        # Stage 1 → vector search
        docs = chroma.similarity_search(query, k=k)

        # Stage 2 → rerank
        docs = reranker.call_rank(query, docs)

        return [d.page_content for d in docs]

    return pipeline
```

```python
# 1. Constrói o pipeline (sua função já está correta com o caminho ../RAG/db)
pipeline = build_pipeline()

# 2. Instancia o Evaluator
# O modelo BAAI/bge-m3 será carregado aqui para calcular a similaridade de 40%
evaluator = RAGEvaluator(
    pipeline_fn=pipeline, 
    dataset=dataset,
    model_name="BAAI/bge-m3" 
)

# 3. Roda a avaliação
# k=12 é o que o retriever (Chroma) busca antes do Reranker filtrar para top_n=6
# similarity_threshold=0.40 é o seu critério de hit semântico
evaluator.evaluate(k=12, similarity_threshold=0.75)
```
